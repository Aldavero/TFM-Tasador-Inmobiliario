```python
import os
import shutil

# Creamos las carpetas originales que Windows borró
os.makedirs('data_pipeline/data/processed', exist_ok=True)
os.makedirs('data_pipeline/data/raw', exist_ok=True)

# Movemos el excel a su sitio
if os.path.exists('propiedades_etiquetadas.csv'):
    shutil.move('propiedades_etiquetadas.csv', 'data_pipeline/data/processed/propiedades_etiquetadas.csv')

# Movemos la carpeta de fotos a su sitio
if os.path.exists('images'):
    shutil.move('images', 'data_pipeline/data/raw/images')

print("¡Archivos colocados correctamente!")
```

```python
import pandas as pd

# Abrimos tu Excel
csv_path = 'data_pipeline/data/processed/propiedades_etiquetadas.csv'
df = pd.read_csv(csv_path)

# Le damos la vuelta a todas las barras de las fotos para que Linux las entienda
columnas_fotos = ['local_img_1', 'local_img_2', 'local_img_3', 'local_img_4', 'local_img_5']
for col in columnas_fotos:
    df[col] = df[col].astype(str).str.replace('\\', '/')

# Guardamos el Excel corregido
df.to_csv(csv_path, index=False)

print("¡Excel arreglado para Linux!")
```
